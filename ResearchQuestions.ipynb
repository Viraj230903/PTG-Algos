RQ1 — Performance

Section 9 of the literature review identifies six criteria a game terrain generator must satisfy, and observes that no single algorithm meets all of them. Speed, memory footprint, and scaling behaviour underpin three of these criteria — real-time viability, seed-consistent extensibility, and deployment in large open worlds. Yet the reviewed papers report these properties only informally, often on incompatible hardware and rarely across resolution tiers. RQ1 therefore asks: how do noise-based, simulation-augmented, and constraint-based approaches compare in generation time, memory footprint, and scaling behaviour under controlled conditions?

RQ2 — Visual quality

Sections 3 and 4 of the literature review together frame the central visual-quality claim of the field: noise-based methods produce output that "appears unnaturally smooth" (Section 3) and lacks "large-scale geological structure", while erosion simulation "consistently improves perceived realism" (Section 4). Both claims are widely asserted but rarely measured directly, a symptom of the evaluation gap identified in Section 9. RQ2 therefore asks: to what extent do the four dimensions of terrain realism — surface roughness variation, height-distribution asymmetry, slope-angle characteristics, and drainage coherence — differ quantitatively between the three paradigms, and does erosion measurably close the gap between raw noise and structurally realistic terrain?

RQ3 — Controllability

Section 5 highlights constraint-based methods (WFC) as offering strong designer control at the cost of scalability, while noise-based methods offer little direct spatial control despite their production dominance. Section 9 lists "providing designers with high-level controls" as one of three open problems in the field. Existing work characterises the controllability of each paradigm qualitatively — as a design property — but rarely measures how successfully each paradigm can be steered toward specified terrain features. RQ3 therefore asks: given a fixed set of designer-authored terrain specifications, what fraction can each algorithm satisfy, at what cost to output quality and generation time?
